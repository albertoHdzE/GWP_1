```python
import numpy as np
```

# CoVaR

```python
"""PSEUDOCODE:
Estimate institution VaR at confidence 
Estimate system loss conditional on institution = VaR 
Return conditional system VaR (CoVaR)"""

# Code
def covar(s,i,q=.05):
    v=np.quantile(i,q)
    return np.quantile(s[i<=v],q)

# Test
s=np.array([-.1,-.02,.01,-.08,.03])
i=np.array([-.2,-.01,.02,-.1,.04])
covar(s,i)
```

# Co-Risk

```python
"""PSEUDOCODE:
Estimate institution distress level
Estimate expected loss of another institution/system
Condition on first institution being distressed
Return conditional expected loss"""

# Code
def corisk(s,i,q=.05):
    t=np.quantile(i,q)
    return s[i<=t].mean()

# Test
s=np.array([-.1,-.02,.01,-.08,.03])
i=np.array([-.2,-.01,.02,-.1,.04])
corisk(s,i)
```

# Marginal Expected Shortfall

```python
"""PSEUDOCODE:
Identify days when market return is below VaR threshold
Compute average institution return on those days
Return MES"""

# Code
def mes(m,f,q=.05):
    v=np.quantile(m,q)
    return f[m<=v].mean()

# Test
m=np.array([-.1,-.03,.01,-.08,.02])
f=np.array([-.15,-.02,.03,-.1,.01])
mes(m,f)
```
