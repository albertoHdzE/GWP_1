### Hamilton Framework Features:

- **Regime-dependent autoregressive coefficients**: φ₀(sₜ), φ₁(sₜ), ..., φₚ(sₜ)
- **Regime-dependent error variance**: σ²(sₜ)
- **Markov-switching transition probabilities**: P(sₜ = j | sₜ₋₁ = i)
- **Optimal filtering and smoothing**: Forward-backward algorithm for regime inference

The general model specification is:

yₜ = φ₀(sₜ) + φ₁(sₜ)yₜ₋₁ + ... + φₚ(sₜ)yₜ₋ₚ + εₜ

where εₜ ~ N(0, σ²(sₜ)) and sₜ follows a Markov chain.
