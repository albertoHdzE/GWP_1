```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
S0 = 80
r = 0.055
v0 = 0.032
kappa = 1.85
theta = 0.045
sigma = 0.35
T = 0.25
rho = -0.30
K = 80  # ATM strike
n_simulations = 100000
n_steps = 252
dt = T / n_steps

# Heston Model Monte-Carlo Simulation
def heston_simulation(S0, v0, r, kappa, theta, sigma, rho, T, n_steps, n_simulations):
    np.random.seed(42)
    dt = T / n_steps
    S = np.zeros((n_simulations, n_steps + 1))
    v = np.zeros((n_simulations, n_steps + 1))
    S[:, 0] = S0
    v[:, 0] = v0
    
    # Correlated Brownian motions
    Z_v = np.random.normal(0, 1, (n_simulations, n_steps))
    Z_s = rho * Z_v + np.sqrt(1 - rho**2) * np.random.normal(0, 1, (n_simulations, n_steps))
    
    for t in range(1, n_steps + 1):
        # Variance process (CIR model)
        v[:, t] = np.maximum(v[:, t-1] + kappa * (theta - v[:, t-1]) * dt + 
                             sigma * np.sqrt(np.maximum(v[:, t-1], 0)) * np.sqrt(dt) * Z_v[:, t-1], 0)
        # Stock price process
        S[:, t] = S[:, t-1] * np.exp((r - 0.5 * v[:, t-1]) * dt + 
                                     np.sqrt(np.maximum(v[:, t-1], 0)) * np.sqrt(dt) * Z_s[:, t-1])
    
    return S, v

# Pricing function
def price_heston_options(S0, K, r, T, n_simulations, n_steps, v0, kappa, theta, sigma, rho):
    S, _ = heston_simulation(S0, v0, r, kappa, theta, sigma, rho, T, n_steps, n_simulations)
    
    # Payoffs
    call_payoff = np.maximum(S[:, -1] - K, 0)
    put_payoff = np.maximum(K - S[:, -1], 0)
    
    # Discounted expected payoff
    call_price = np.exp(-r * T) * np.mean(call_payoff)
    put_price = np.exp(-r * T) * np.mean(put_payoff)
    
    return call_price, put_price

# Run simulation
call_price, put_price = price_heston_options(S0, K, r, T, n_simulations, n_steps, v0, kappa, theta, sigma, rho)

# Plot sample paths
S, _ = heston_simulation(S0, v0, r, kappa, theta, sigma, rho, T, n_steps, 10)  # 10 paths for visualization
plt.figure(figsize=(10, 6))
for i in range(10):
    plt.plot(np.linspace(0, T, n_steps + 1), S[i, :], label=f'Path {i+1}' if i < 5 else None)
plt.title('Sample Stock Price Paths (Heston Model, Correlation = -0.30)')
plt.xlabel('Time (Years)')
plt.ylabel('Stock Price')
plt.grid(True)
plt.savefig('/Users/beto/Documents/Projects/WQU/03-DerivativePricing/GWP-2/report/images/q1.png')
plt.close()

# Print results for reference
print(f"ATM European Call Price: {call_price:.2f}")
print(f"ATM European Put Price: {put_price:.2f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
S0 = 80
r = 0.055
v0 = 0.032
kappa = 1.85
theta = 0.045
sigma = 0.35
T = 0.25
K = 80  # ATM strike
n_simulations = 100000
n_steps = 252
dt = T / n_steps
rhos = [-0.30, -0.70]
delta_S = 0.1  # Perturbation for numerical derivatives

# Heston Model Monte-Carlo Simulation
def heston_simulation(S0, v0, r, kappa, theta, sigma, rho, T, n_steps, n_simulations):
    np.random.seed(42)
    dt = T / n_steps
    S = np.zeros((n_simulations, n_steps + 1))
    v = np.zeros((n_simulations, n_steps + 1))
    S[:, 0] = S0
    v[:, 0] = v0
    
    Z_v = np.random.normal(0, 1, (n_simulations, n_steps))
    Z_s = rho * Z_v + np.sqrt(1 - rho**2) * np.random.normal(0, 1, (n_simulations, n_steps))
    
    for t in range(1, n_steps + 1):
        v[:, t] = np.maximum(v[:, t-1] + kappa * (theta - v[:, t-1]) * dt + 
                             sigma * np.sqrt(np.maximum(v[:, t-1], 0)) * np.sqrt(dt) * Z_v[:, t-1], 0)
        S[:, t] = S[:, t-1] * np.exp((r - 0.5 * v[:, t-1]) * dt + 
                                     np.sqrt(np.maximum(v[:, t-1], 0)) * np.sqrt(dt) * Z_s[:, t-1])
    
    return S, v

# Pricing function
def price_heston_options(S0, K, r, T, n_simulations, n_steps, v0, kappa, theta, sigma, rho):
    S, _ = heston_simulation(S0, v0, r, kappa, theta, sigma, rho, T, n_steps, n_simulations)
    call_payoff = np.maximum(S[:, -1] - K, 0)
    put_payoff = np.maximum(K - S[:, -1], 0)
    call_price = np.exp(-r * T) * np.mean(call_payoff)
    put_price = np.exp(-r * T) * np.mean(put_payoff)
    return call_price, put_price

# Calculate Greeks
results = []
for rho in rhos:
    # Base prices
    call_price, put_price = price_heston_options(S0, K, r, T, n_simulations, n_steps, v0, kappa, theta, sigma, rho)
    # Perturbed prices
    call_price_up, put_price_up = price_heston_options(S0 + delta_S, K, r, T, n_simulations, n_steps, v0, kappa, theta, sigma, rho)
    call_price_down, put_price_down = price_heston_options(S0 - delta_S, K, r, T, n_simulations, n_steps, v0, kappa, theta, sigma, rho)
    
    # Delta
    call_delta = (call_price_up - call_price_down) / (2 * delta_S)
    put_delta = (put_price_up - put_price_down) / (2 * delta_S)
    
    # Gamma
    call_gamma = (call_price_up - 2 * call_price + call_price_down) / (delta_S ** 2)
    put_gamma = (put_price_up - 2 * put_price + put_price_down) / (delta_S ** 2)
    
    results.append((rho, call_price, put_price, call_delta, put_delta, call_gamma, put_gamma))

# Plot option prices vs stock price
S_range = np.linspace(S0 - 5, S0 + 5, 50)
call_prices_rho1 = []
call_prices_rho2 = []
for S in S_range:
    c1, _ = price_heston_options(S, K, r, T, n_simulations, n_steps, v0, kappa, theta, sigma, rhos[0])
    c2, _ = price_heston_options(S, K, r, T, n_simulations, n_steps, v0, kappa, theta, sigma, rhos[1])
    call_prices_rho1.append(c1)
    call_prices_rho2.append(c2)

plt.figure(figsize=(10, 6))
plt.plot(S_range, call_prices_rho1, label='Correlation = -0.30')
plt.plot(S_range, call_prices_rho2, label='Correlation = -0.70')
plt.title('Call Option Prices vs. Stock Price (Heston Model)')
plt.xlabel('Stock Price')
plt.ylabel('Option Price')
plt.legend()
plt.grid(True)
plt.savefig('/Users/beto/Documents/Projects/WQU/03-DerivativePricing/GWP-2/report/images/q3.png')
plt.close()

# Print results
for rho, c_price, p_price, c_delta, p_delta, c_gamma, p_gamma in results:
    print(f"Correlation: {rho}")
    print(f"Call Price: {c_price:.2f}, Delta: {c_delta:.2f}, Gamma: {c_gamma:.3f}")
    print(f"Put Price: {p_price:.2f}, Delta: {p_delta:.2f}, Gamma: {p_gamma:.3f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# General Parameters
S0 = 80
r = 0.055
T = 0.25
sigma = 0.35
n_simulations = 100000
n_steps = 252
dt = T / n_steps

# Heston Parameters
v0 = 0.032
kappa = 1.85
theta = 0.045
rho = -0.30

# Merton Parameters
mu_j = -0.5
delta = 0.22
lambda_j = 0.75

# Strike prices based on moneyness
moneyness = [0.85, 0.90, 0.95, 1.00, 1.05, 1.10, 1.15]
strikes = [S0 / m for m in moneyness]

# Heston Model Monte-Carlo Simulation
def heston_simulation(S0, v0, r, kappa, theta, sigma, rho, T, n_steps, n_simulations):
    np.random.seed(42)
    dt = T / n_steps
    S = np.zeros((n_simulations, n_steps + 1))
    v = np.zeros((n_simulations, n_steps + 1))
    S[:, 0] = S0
    v[:, 0] = v0
    
    Z_v = np.random.normal(0, 1, (n_simulations, n_steps))
    Z_s = rho * Z_v + np.sqrt(1 - rho**2) * np.random.normal(0, 1, (n_simulations, n_steps))
    
    for t in range(1, n_steps + 1):
        v[:, t] = np.maximum(v[:, t-1] + kappa * (theta - v[:, t-1]) * dt + 
                             sigma * np.sqrt(np.maximum(v[:, t-1], 0)) * np.sqrt(dt) * Z_v[:, t-1], 0)
        S[:, t] = S[:, t-1] * np.exp((r - 0.5 * v[:, t-1]) * dt + 
                                     np.sqrt(np.maximum(v[:, t-1], 0)) * np.sqrt(dt) * Z_s[:, t-1])
    
    return S

# Merton Model Monte-Carlo Simulation
def merton_simulation(S0, r, sigma, mu_j, delta, lambda_j, T, n_steps, n_simulations):
    np.random.seed(42)
    dt = T / n_steps
    S = np.zeros((n_simulations, n_steps + 1))
    S[:, 0] = S0
    
    jump_adj = np.exp(mu_j + 0.5 * delta**2) - 1
    drift = r - 0.5 * sigma**2 - lambda_j * jump_adj
    
    for t in range(1, n_steps + 1):
        Z = np.random.normal(0, 1, n_simulations)
        N = np.random.poisson(lambda_j * dt, n_simulations)
        J = np.array([np.sum(np.random.normal(mu_j, delta, n)) if n > 0 else 0 for n in N])
        S[:, t] = S[:, t-1] * np.exp(drift * dt + sigma * np.sqrt(dt) * Z + J)
    
    return S

# Pricing function for call options
def price_call_option(S, K, r, T):
    call_payoff = np.maximum(S[:, -1] - K, 0)
    call_price = np.exp(-r * T) * np.mean(call_payoff)
    return call_price

# Compute prices for all strikes
heston_prices = []
merton_prices = []
for K in strikes:
    # Heston
    S_heston = heston_simulation(S0, v0, r, kappa, theta, sigma, rho, T, n_steps, n_simulations)
    heston_price = price_call_option(S_heston, K, r, T)
    heston_prices.append(heston_price)
    
    # Merton
    S_merton = merton_simulation(S0, r, sigma, mu_j, delta, lambda_j, T, n_steps, n_simulations)
    merton_price = price_call_option(S_merton, K, r, T)
    merton_prices.append(merton_price)

# Print results
print("Heston Model Call Prices:")
for m, K, price in zip(moneyness, strikes, heston_prices):
    print(f"Moneyness: {m:.2f}, Strike: {K:.2f}, Call Price: {price:.2f}")

print("\nMerton Model Call Prices:")
for m, K, price in zip(moneyness, strikes, merton_prices):
    print(f"Moneyness: {m:.2f}, Strike: {K:.2f}, Call Price: {price:.2f}")

# Plot call prices vs. strike prices
plt.figure(figsize=(10, 6))
plt.plot(strikes, heston_prices, label='Heston (ρ = -0.30)', marker='o')
plt.plot(strikes, merton_prices, label='Merton (λ = 0.75)', marker='s')
plt.title('European Call Option Prices vs. Strike Price')
plt.xlabel('Strike Price')
plt.ylabel('Call Option Price')
plt.legend()
plt.grid(True)
plt.savefig('/Users/beto/Documents/Projects/WQU/03-DerivativePricing/GWP-2/report/images/q8.png')
plt.close()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Parameters
S0 = 80
r = 0.055
v0 = 0.032
kappa = 1.85
theta = 0.045
sigma = 0.35
rho = -0.30
T = 0.25
K = 80  # ATM strike
n_simulations = 100000
n_steps = 252
dt = T / n_steps
delta_S = 0.1  # Perturbation for Greeks

# Heston Model Monte-Carlo Simulation
def heston_simulation(S0, v0, r, kappa, theta, sigma, rho, T, n_steps, n_simulations):
    np.random.seed(42)
    dt = T / n_steps
    S = np.zeros((n_simulations, n_steps + 1))
    v = np.zeros((n_simulations, n_steps + 1))
    S[:, 0] = S0
    v[:, 0] = v0
    
    Z_v = np.random.normal(0, 1, (n_simulations, n_steps))
    Z_s = rho * Z_v + np.sqrt(1 - rho**2) * np.random.normal(0, 1, (n_simulations, n_steps))
    
    for t in range(1, n_steps + 1):
        v[:, t] = np.maximum(v[:, t-1] + kappa * (theta - v[:, t-1]) * dt + 
                             sigma * np.sqrt(np.maximum(v[:, t-1], 0)) * np.sqrt(dt) * Z_v[:, t-1], 0)
        S[:, t] = S[:, t-1] * np.exp((r - 0.5 * v[:, t-1]) * dt + 
                                     np.sqrt(np.maximum(v[:, t-1], 0)) * np.sqrt(dt) * Z_s[:, t-1])
    
    return S, v

# American Call Pricing with Longstaff-Schwartz
def price_american_call(S, K, r, dt):
    n_simulations, n_steps_plus1 = S.shape
    n_steps = n_steps_plus1 - 1
    payoff = np.zeros((n_simulations, n_steps + 1))
    payoff[:, -1] = np.maximum(S[:, -1] - K, 0)
    
    for t in range(n_steps - 1, -1, -1):
        in_the_money = S[:, t] > K
        if np.any(in_the_money):
            X = np.vstack([S[in_the_money, t], S[in_the_money, t]**2, np.ones(np.sum(in_the_money))]).T
            Y = np.exp(-r * dt * (n_steps - t)) * payoff[in_the_money, t + 1]
            reg = LinearRegression().fit(X, Y)
            continuation = reg.predict(X)
            exercise = np.maximum(S[in_the_money, t] - K, 0)
            payoff[in_the_money, t] = np.where(exercise > continuation, exercise, np.exp(-r * dt) * payoff[in_the_money, t + 1])
            payoff[~in_the_money, t] = np.exp(-r * dt) * payoff[~in_the_money, t + 1]
        else:
            payoff[:, t] = np.exp(-r * dt) * payoff[:, t + 1]
    
    return np.mean(payoff[:, 0])

# Compute price and Greeks
def price_american_call_and_greeks(S0, K, r, T, n_simulations, n_steps, v0, kappa, theta, sigma, rho, delta_S):
    # Base price
    S, _ = heston_simulation(S0, v0, r, kappa, theta, sigma, rho, T, n_steps, n_simulations)
    price = price_american_call(S, K, r, T / n_steps)
    
    # Perturbed prices for Greeks
    S_up, _ = heston_simulation(S0 + delta_S, v0, r, kappa, theta, sigma, rho, T, n_steps, n_simulations)
    price_up = price_american_call(S_up, K, r, T / n_steps)
    S_down, _ = heston_simulation(S0 - delta_S, v0, r, kappa, theta, sigma, rho, T, n_steps, n_simulations)
    price_down = price_american_call(S_down, K, r, T / n_steps)
    
    # Delta
    delta = (price_up - price_down) / (2 * delta_S)
    
    # Gamma
    gamma = (price_up - 2 * price + price_down) / (delta_S ** 2)
    
    return price, delta, gamma

# Run simulation
call_price, call_delta, call_gamma = price_american_call_and_greeks(S0, K, r, T, n_simulations, n_steps, v0, kappa, theta, sigma, rho, delta_S)

# Plot sample paths
S, _ = heston_simulation(S0, v0, r, kappa, theta, sigma, rho, T, n_steps, 10)  # 10 paths for visualization
plt.figure(figsize=(10, 6))
for i in range(10):
    plt.plot(np.linspace(0, T, n_steps + 1), S[i, :], label=f'Path {i+1}' if i < 5 else None)
plt.title('Sample Stock Price Paths (Heston Model, American Call, ρ = -0.30)')
plt.xlabel('Time (Years)')
plt.ylabel('Stock Price')
plt.grid(True)
plt.savefig('/Users/beto/Documents/Projects/WQU/03-DerivativePricing/GWP-2/report/images/q9.png')
plt.close()

# Print results
print(f"American Call Price: {call_price:.2f}")
print(f"Delta: {call_delta:.2f}")
print(f"Gamma: {call_gamma:.3f}")
```
